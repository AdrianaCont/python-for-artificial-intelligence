```python
import pandas as pd
#urlInventory = "https://raw.githubusercontent.com/AdrianaCont/python-for-artificial-intelligence/refs/heads/main/Datasets/Inventory_dataset.csv"
urlSales = "https://raw.githubusercontent.com/AdrianaCont/python-for-artificial-intelligence/refs/heads/main/Datasets/Sales_dataset.csv"
```

```python
df_sales = pd.read_csv(urlSales)

# isna() -> ¿qué valores estan auscentes? true = auscente, false= si existe valor
df_sales.isna()

# queremos saber, cuántos valores faltan por columna
df_sales.isna().sum()

```

```python
# nunique() -> ¿cuántos valores únicos hay por columna?
df_sales.nunique() #number of unique values

df_sales["tx_Distribuidor"].nunique()

df_sales["cd_IdClienteFC"].nunique()
```

```python
# cuántas veces aparece cada valor?
df_sales["tx_Vendedor"].value_counts()
df_sales["ct_Cantidad"].value_counts() # aquí aparecen 47 valores de cantidad en 0 ¿porqué?
```

```python
#duplicated() ¿Tengo filas repetidas?
df_sales.duplicated()
#Cuántas filas completas están duplicadas?
df_sales.duplicated().sum()

df_sales["ct_Cantidad"].describe() #no se está describiendo todo, solo ct_cantidad
```
