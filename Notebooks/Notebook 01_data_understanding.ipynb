## Granularidad

La ganularidad normalmente se puede escribir como:

Una fila por _______________ .

Por ejemplo:
Una fila por cliente
Una fila por factura
Una fila por producto
Una fila por producto-factura
Una fila por cliente-mes
Una fila por dealer-producto-mes

Cuantas más dimensiones se agregan, normalmente se obitene mayor detalle.
