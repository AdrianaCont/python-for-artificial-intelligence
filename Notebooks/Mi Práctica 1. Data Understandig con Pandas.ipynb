```python
import pandas as pd
urlInventory = "https://raw.githubusercontent.com/AdrianaCont/python-for-artificial-intelligence/refs/heads/main/Datasets/Inventory_dataset.csv"
urlSales = "https://raw.githubusercontent.com/AdrianaCont/python-for-artificial-intelligence/refs/heads/main/Datasets/Sales_dataset.csv"
```

```python
df_inventory = pd.read_csv(urlInventory)
df_sales = pd.read_csv(urlSales)
print(df_sales.head()); # primeras 5 filas
print(df_sales.shape);  # No filas y columnas
print(df_sales.columns); # Cols
print(df_sales.dtypes);  # Tipos
print(df_sales.describe()); # Estadisticas
print(df_inventory.describe(include='all')); # Estadisticas
print(df_sales.info()); # Info
print(df_inventory.isnull().sum()); # 
```
